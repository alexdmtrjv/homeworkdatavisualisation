# Homework 5 — Data Visualisation with Python

**Dataset:** Languages of the World

## Step 1: Load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('languages_world (1).csv')
df.head()
```

## Step 2: Explore the data

```python
print('Rows:', len(df))
print('Columns:', len(df.columns))
df
```

## Visualisation 1 — Number of Speakers

A bar chart compares the number of speakers across languages.

```python
x = df.sort_values('Speakers_millions', ascending=False)
plt.figure(figsize=(12,7))
plt.bar(x['Language'], x['Speakers_millions'])
plt.xticks(rotation=75, ha='right')
plt.xlabel('Language')
plt.ylabel('Speakers (millions)')
plt.title('Number of Speakers by Language')
plt.tight_layout()
plt.show()
```

**Reflection:** English has the highest number of speakers in this dataset. Mandarin and Hindi also have very large speaker populations, while Swedish and Dutch have much smaller values.

## Visualisation 2 — Languages by Language Family

A pie chart shows the distribution of languages among language families.

```python
counts = df['Language_family'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%')
plt.title('Languages by Language Family')
plt.tight_layout()
plt.show()
```

**Reflection:** Indo-European is the largest language family in the dataset. The remaining families occur much less frequently.

## Visualisation 3 — Speakers by Language Family

A bar chart compares the total number of speakers represented by each language family.

```python
totals = df.groupby('Language_family')['Speakers_millions'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,6))
plt.bar(totals.index, totals.values)
plt.xticks(rotation=45, ha='right')
plt.xlabel('Language family')
plt.ylabel('Total speakers (millions)')
plt.title('Total Speakers by Language Family')
plt.tight_layout()
plt.show()
```

**Reflection:** Indo-European has by far the largest total number of speakers in this dataset. This is partly because the dataset contains many Indo-European languages and several of them have large speaker populations.

## Conclusion

The visualisations show clear differences between languages and language families. The dataset is especially dominated by Indo-European languages and by languages with very large speaker populations.
